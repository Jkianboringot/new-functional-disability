```python

import cufflinks as cf
import chart_studio.plotly as py
import seaborn as sns
import plotly.express as px
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.offline import download_plotlyjs, init_notebook_mode, plot, iplot
import warnings
warnings.filterwarnings("ignore")
init_notebook_mode(connected=True)
cf.go_offline()
```

```python
df=pd.read_csv("keep/csv/final_FDD.csv")
df.head()
```

## line charts

```python
disability = (
    df.loc[(df["Sex"] != "Both Sexes")& (df["Status"] != "All") ]
    .groupby(["Disability", "Region","Sex"], 
             as_index=False)["Household Population 5 Years Old and Over with Functional Difficulty"].sum())
figpx=px.line(disability,x="Sex",y="Household Population 5 Years Old and Over with Functional Difficulty"
            ,color='Disability',markers=True)
#fig.update_traces(textposition="bottomright")
figpx
```

```python



figgo=go.Figure()
figgo.add_traces(go.Scatter(x=disability["Region"],
                                y=disability["Household Population 5 Years Old and Over with Functional Difficulty"],
                                mode="lines",name="LInes")
                                )
figgo.add_traces(go.Scatter(x=disability["Disability"],
                                y=disability["Household Population 5 Years Old and Over with Functional Difficulty"],
                                mode="lines+markers",name="LInes")
                                )
figgo.add_traces(go.Scatter(x=disability["Sex"],
                                y=disability["Household Population 5 Years Old and Over with Functional Difficulty"],
                                mode="markers",name="LInes")
                                )
figgo.update_layout(
    xaxis=dict(
        showline=True,
        showgrid=True,
        showticklabels=True,
        linecolor="rgb(204,204,204)",
        linewidth=1,
        ticks="outside",
        tickfont=dict(
            family="Arial",
            size=12,
            color="rgb(82,82,82)"
        )
    ),
    yaxis=dict(
        showgrid=False,
        zeroline=False,
        showline=True,
        showticklabels=False
    ),
    autosize=True,
    margin=dict(
        autoexpand=False,
        l=20,
        r=20,
        t=1
    ),
    showlegend=True,
    plot_bgcolor="#F0F0F0"
)


figgo.show()
```

## bar chart

```python
bar_disability = (
    df.loc[(df["Sex"] != "Both Sexes")& (df["Status"] != "All") ]
    .groupby(["Region"], 
             as_index=False)["Household Population 5 Years Old and Over with Functional Difficulty"].sum())
```

```python
px.bar(disability,x="Disability",y="Household Population 5 Years Old and Over with Functional Difficulty",
       color="Sex",title="disability by sex",
       labels={"Disability":"Disability type",
               "Household Population 5 Years Old and Over with Functional Difficulty":"Disability count"}
               ,barmode="group")





fig=px.bar(bar_disability,x="Region",y="Household Population 5 Years Old and Over with Functional Difficulty",
       color="Region",title="disability by sex",
       labels={"Disability":"Disability type",
               "Household Population 5 Years Old and Over with Functional Difficulty":"Disability count"},
               text="Household Population 5 Years Old and Over with Functional Difficulty"
               )
fig.update_traces(texttemplate="%{text:.2s}",textposition="outside")
fig.update_layout(uniformtext_minsize=10, uniformtext_mode="hide")
fig.update_layout(margin=dict(autoexpand=False,l=50,r=200,t=30),autosize=True,showlegend=True,
                   plot_bgcolor="#F0F0F0")
fig
```

## Scatter plot

```python
scatter_disability = (
    df.loc[(df["Sex"] != "Both Sexes")& (df["Status"] != "All") ]
    .groupby(["Region"], 
             as_index=False)["Household Population 5 Years Old and Over with Functional Difficulty"].sum())
```

```python
k=px.scatter(bar_disability,x="Region",y="Household Population 5 Years Old and Over with Functional Difficulty"
           ,color="Region",size="Household Population 5 Years Old and Over with Functional Difficulty",
           hover_data="Household Population 5 Years Old and Over with Functional Difficulty"
           ,hover_name="Region")
k.update_layout(uniformtext_minsize=10, uniformtext_mode="hide")
k.update_layout(margin=dict(autoexpand=False,l=50,r=200,t=30),autosize=True,showlegend=True,
                   plot_bgcolor="#F0F0F0")
```

```python
#this can be actually be good enough to put in website
prov = (
    df.loc[(df["Sex"] != "Both Sexes")& (df["Status"] != "All") ]
    .groupby(["Province"], 
             as_index=False)["Household Population 5 Years Old and Over with Functional Difficulty"].sum())

figsc=go.Figure()
figsc.add_trace(go.Scatter(x=prov["Province"],
                           y=prov["Household Population 5 Years Old and Over with Functional Difficulty"],
                           mode="markers",  marker_color=prov["Household Population 5 Years Old and Over with Functional Difficulty"],
                           marker=dict(showscale=True ,colorscale='Viridis')))
figsc.update_traces(marker_line_width=1,marker_size=16)
figsc.update_layout(title=dict(text="Scatter plot by region disability"))  
figsc.update_layout(
    margin=dict(autoexpand=False,r=100,l=50,t=50),showlegend=False,autosize=True
                )
figsc
```

```python

fig=px.bar(prov,x="Province",y="Household Population 5 Years Old and Over with Functional Difficulty",
       color="Province",title="disability by sex",
       labels={"Disability":"Disability type",
               "Household Population 5 Years Old and Over with Functional Difficulty":"Disability count"},
               text="Household Population 5 Years Old and Over with Functional Difficulty"
               )
fig.update_traces(texttemplate="%{text:.2s}",textposition="outside")
fig.update_layout(uniformtext_minsize=10, uniformtext_mode="hide")
fig.update_layout(margin=dict(autoexpand=False,l=50,r=200,t=30),autosize=True,showlegend=True,
                   plot_bgcolor="#F0F0F0")
```

## box plot


```python
sex = (
    df.loc[(df["Sex"] != "Both Sexes")& (df["Status"] != "All") ]
    .groupby(["Sex"], 
             as_index=False)["Household Population 5 Years Old and Over with Functional Difficulty"].sum())

px.box(disability,x="Sex",y="Household Population 5 Years Old and Over with Functional Difficulty",points="all",color="Sex")

fig=go.Figure()
fig.add_traces(go.Box(x=disability["Sex"],y=disability["Household Population 5 Years Old and Over with Functional Difficulty"],
                      boxpoints="suspectedoutliers",marker=dict(color="red",outliercolor='red'),
    line_color='rgb(8,81,156)',
                      jitter=0.6,marker_color="blue",fillcolor="cyan",whiskerwidth=0.5))
fig.update_layout(title="WHisker plot",yaxis=dict(gridcolor="black",gridwidth=1),paper_bgcolor="gray",plot_bgcolor="gray")
```
